```python
import requests
import pandas as pd
import os

# Direct download link (reliable CSV)
url = "https://data.humdata.org/dataset/3b4a119a-309c-4d3f-900f-18a1f6ca2dfa/resource/4658aa59-0554-4fac-8473-377da4b7a0e9/download/nigeriahealthfacilities.csv"

# Folder where the file will be saved
save_folder = r"C:\Users\USER\Downloads"
os.makedirs(save_folder, exist_ok=True)

file_path = os.path.join(save_folder, "nigeria_health_facilities.csv")

print("Downloading dataset... Please wait.")

# Download the file
response = requests.get(url, stream=True)
response.raise_for_status()   # Check if download was successful

with open(file_path, "wb") as f:
    for chunk in response.iter_content(chunk_size=8192):
        f.write(chunk)

print("Download completed successfully!")
print(f"File saved at: {file_path}")

# Load the dataset
df = pd.read_csv(file_path)

print("\nDataset loaded successfully!")
print("Shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())
df.head()
```

```python
print("Dataset Shape:", df.shape)
print("\nColumn Names:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

print("\nFirst 5 rows:")
df.head()
```

```python
print("Missing Values:")
print(df.isnull().sum())

print("\nPercentage of Missing Values:")
print((df.isnull().sum() / len(df) * 100).round(2))

print("\nNumber of duplicate rows:", df.duplicated().sum())
```

```python
print("Functional Status:")
print(df['functional_status'].value_counts())

print("\nType:")
print(df['type'].value_counts())

print("\nCategory (Top 15):")
print(df['category'].value_counts().head(15))

print("\nState Name (Top 15):")
print(df['state_name'].value_counts().head(15))
```

```python
# Fill missing category with 'Unknown'
df['category'] = df['category'].fillna('Unknown')

# Create a cleaner functional status column if needed
print("\nCleaned Category missing values:", df['category'].isnull().sum())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 6))
ax = sns.countplot(data=df, x='functional_status', order=df['functional_status'].value_counts().index, palette='Set2')

plt.title('Distribution of Health Facilities by Functional Status', fontsize=14)
plt.xlabel('Functional Status')
plt.ylabel('Number of Facilities')
plt.xticks(rotation=15)

# Add value labels
for p in ax.patches:
    ax.annotate(f'{int(p.get_height()):,}', 
                (p.get_x() + p.get_width() / 2., p.get_height()),
                ha='center', va='bottom', fontsize=10)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))
ax = sns.countplot(data=df, y='type', order=df['type'].value_counts().index, palette='viridis')

plt.title('Distribution of Health Facilities by Type', fontsize=14)
plt.xlabel('Number of Facilities')
plt.ylabel('Facility Type')

for p in ax.patches:
    width = p.get_width()
    ax.text(width + 200, p.get_y() + p.get_height()/2, f'{int(width):,}', va='center')

plt.tight_layout()
plt.show()
```

```python
top_states = df['state_name'].value_counts().head(15)

plt.figure(figsize=(12, 7))
ax = sns.barplot(x=top_states.values, y=top_states.index, palette='coolwarm')

plt.title('Top 15 States by Number of Health Facilities', fontsize=14)
plt.xlabel('Number of Facilities')
plt.ylabel('State')

for i, v in enumerate(top_states.values):
    ax.text(v + 100, i, f'{v:,}', va='center')

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 6))
sns.countplot(data=df, x='type', hue='functional_status', 
              order=df['type'].value_counts().index[:5],
              palette='Set2')

plt.title('Functional Status by Facility Type', fontsize=14)
plt.xlabel('Facility Type')
plt.ylabel('Number of Facilities')
plt.xticks(rotation=20)
plt.legend(title='Functional Status', bbox_to_anchor=(1.05, 1))
plt.tight_layout()
plt.show()
```

```python
functional_rate = df.groupby('type')['functional_status'].apply(
    lambda x: (x == 'Functional').mean() * 100
).sort_values(ascending=False).round(1)

print("Percentage of Functional Facilities by Type:")
print(functional_rate)

plt.figure(figsize=(10, 6))
ax = sns.barplot(x=functional_rate.values, y=functional_rate.index, palette='Greens')

plt.title('Percentage of Functional Facilities by Type', fontsize=14)
plt.xlabel('Percentage Functional (%)')
plt.ylabel('Facility Type')

for i, v in enumerate(functional_rate.values):
    ax.text(v + 0.5, i, f'{v}%', va='center')

plt.tight_layout()
plt.show()
```

```python
# Create a simplified status
df['status_simple'] = df['functional_status'].replace({
    'Partially Functional': 'Not Fully Functional',
    'Not Functional': 'Not Fully Functional',
    'Unknown': 'Unknown'
})

top_10_states = df['state_name'].value_counts().head(10).index
df_top = df[df['state_name'].isin(top_10_states)]

plt.figure(figsize=(13, 7))
sns.countplot(data=df_top, y='state_name', hue='status_simple',
              order=top_10_states, palette='coolwarm')

plt.title('Top 10 States: Functional Status Breakdown', fontsize=14)
plt.xlabel('Number of Facilities')
plt.ylabel('State')
plt.legend(title='Status')
plt.tight_layout()
plt.show()
```

```python
print("Top 10 Categories:")
print(df['category'].value_counts().head(10))

plt.figure(figsize=(12, 7))
top_categories = df['category'].value_counts().head(10)
ax = sns.barplot(x=top_categories.values, y=top_categories.index, palette='magma')

plt.title('Top 10 Health Facility Categories', fontsize=14)
plt.xlabel('Number of Facilities')
plt.ylabel('Category')

for i, v in enumerate(top_categories.values):
    ax.text(v + 50, i, f'{v:,}', va='center')

plt.tight_layout()
plt.show()
```
